# Dataset Understanding Summary

## 1. Dataset Overview
- Dataset: IITA Irrigated Endline Survey: Plastic Crates RCT Experiment in Nigeria.
- Initial shape: 928 rows and 1,949 columns.
- No exact duplicate rows were identified.

## 2. Production Systems
The dataset contains two production systems:
- Irrigated production (IRG): 877 records with at least one recorded loss value.
- Rain-fed production (RF): 132 records with at least one recorded loss value.
- 81 records contain loss observations for both systems.

Irrigated production provides a larger initial population for modelling.

## 3. Loss Variables
The investigated loss stages include:
- Farm loss.
- Harvest loss.
- Storage loss.
- Transportation loss.

Storage and transportation losses have considerably fewer positive observations than farm and harvest losses.

## 4. Data Quality Observations
- Loss variables contain missing values.
- Corresponding unit-description columns are available.
- Summary and detailed harvest-loss records do not consistently match.
- The measurement conventions of the numerical loss values require further verification.

## 5. Preliminary Modelling Direction
Irrigated production will be considered as the primary modelling population because it has more recorded observations.

The final target variable and predictor features will be selected during preprocessing after validating the measurement definitions.

## 6. Outstanding Limitations
The dataset documentation does not yet establish the exact measurement interpretation of every loss variable. Missing values will not automatically be treated as zero, and loss stages will not be combined without justification.
