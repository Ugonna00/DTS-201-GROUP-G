```python
#Age distribution of farmers
plt.figure(figsize=(8, 5))

sns.histplot(df['respAge'], bins=15, kde=True)

plt.title('Distribution of Farmers Age')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.show()
```

```python
#Tomato production distribution
plt.figure(figsize=(8, 5))

sns.histplot(df['totProd_IRG'], bins=30, kde=True)

plt.title('Distribution of Tomato Production')
plt.xlabel('Total Production')
plt.ylabel('Frequency')

plt.show()
```

```python
#Farm loss distribution
plt.figure(figsize=(8, 5))

sns.histplot(df['farmLoss_kg_IRG'], bins=30, kde=True)

plt.title('Distribution of Tomato Farm Losses')
plt.xlabel('Farm Loss')
plt.ylabel('Frequency')

plt.show()
```

```python
#Education level distribution of farmers
plt.figure(figsize=(10, 5))

sns.countplot(data=df, y='respEduLev',
              order=df['respEduLev'].value_counts().index)

plt.title('Educational Level of Farmers')
plt.xlabel('Number of Farmers')
plt.ylabel('Education Level')

plt.show()
```

```python
#Production against farm loss
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='totProd_IRG',
    y='farmLoss_kg_IRG'
)

plt.title('Tomato Production vs Farm Losses')
plt.xlabel('Total Tomato Production')
plt.ylabel('Farm Losses')

plt.show()
```

```python
#Storage practices against farm loss
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x='rcvTRaininStorgePrctc',
    y='farmLoss_kg_IRG'
)

plt.title('Storage Training vs Farm Losses')
plt.xlabel('Received Storage Training')
plt.ylabel('Farm Losses')

plt.show()
```

```python
#Plastic crates usage against farm loss
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x='evrUsdPlstcCrate',
    y='farmLoss_kg_IRG'
)

plt.title('Plastic Crate Usage vs Farm Losses')
plt.xlabel('Ever Used Plastic Crates')
plt.ylabel('Farm Losses')

plt.show()
```

```python
#Correlation heatmap
plt.figure(figsize=(12, 8))

corr = df.select_dtypes(include=np.number).corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')

plt.show()
```

```python
#Outlier detection
Q1 = df['farmLoss_kg_IRG'].quantile(0.25)

Q3 = df['farmLoss_kg_IRG'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

outliers = df[
    (df['farmLoss_kg_IRG'] < lower) |
    (df['farmLoss_kg_IRG'] > upper)
]

print(outliers)
```
